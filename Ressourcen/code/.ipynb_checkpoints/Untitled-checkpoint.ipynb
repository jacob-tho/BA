```python
"""
Variational Problem of the space-time Laplace Equation with periodic b.c. in space and Neumann b.c. in time
Perturbed version with small epsilon > 0
"""
#def update_phi(phi_old, q, mu, r=0.1, epsilon = 1e-10): # Input ändern: fes und mesh außerhalb dieser Funktion definieren
def update_phi(mesh, fes, q, mu, r=0.1, epsilon = 1e-10):
    """
    Todo: Ist das korrekt? Funktion passend umschreiben, dass Argumente und return korrekt sind. Fehlt noch was oder wars das schon?
    """
     
    phi = fes.TrialFunction()
    v = fes.TestFunction() # Testfunktionen und Trial(?) Funktionen, gegen die halt getestet wird

    # Dieses Laplace-Problem hat Form:
    # -r nabla_tx phi = div(mu - r*q))
    # Hier noch leicht pertubieren mit epsilon << 1; Vorzeichen??
 
    F = mu - r*q # Das hier ist rechte Seite
    
    a = BilinearForm(fes) #Bilinear für linke Seite laut Docs
    a += r*(grad(phi)*grad(v)+epsilon*phi*v)*dx
    
    l = LinearForm(fes) #Linear für rechte Seite laut Docs
    l += F * grad(v) * dx # wieso hier noch grad(v) -- partielle Integration! -- Besser als InnerProduct?
    a.Assemble()
    l.Assemble()

    phi_new = GridFunction(fes)
    inv = a.mat.Inverse(fes.FreeDofs(), inverse = "sparsecholesky")
    phi_new.vec.data = inv * l.vec
    #print(phi_new.vec.data) #sind das sinnvolle Werte?
    return phi_new
    
    
    
    """
    # Das hier ist "normale" Laplace Gleichung -> Abändern
    f = CoefficientFunction(1)
    a = BilinearForm(fes)
    a += (grad(u) * grad(v) + epsilon*u*v) * dx

    l = LinearForm(fes)
    l += f * v * dx
    a.Assemble()
    l.Assemble()

    ### Lösung
    gfu = GridFunction(fes)
    inv = a.mat.Inverse(fes.FreeDofs(), inverse = "sparsecholesky")
    gfu.vec.data = inv * l.vec
    """
```

## ALG Update in q

```python
"""
Aus BB(2000)
Solve the convex inf Problem using Newtons Method
nabla_tx(phi) + mu/r = <alpha, beta> = p_n
<a,b> = q_n
solve 
inf{ (a-alpha)^2 + |b-beta|^2 | a+|b|^2/2 \leq 0} 
Mit Newton

Die Menge K ist nicht-leer (offensichtlich) und konvex (auch offensichtlich aber vllt noch beweisen). 
Die Menge ist strikt konvex und koerziv. Die Nebenbedingung ist abg. und konvex + nicht leer
Stelle Lagrangian (bestimme KKT) auf mit 
2(a-alpha) + lambda = 0
2(b-beta) + lambda*b = 0
"""

def newton(alpha, beta, tol=1e-12, maxiter=50):
    if alpha + beta**2/2 <= 0:
        return alpha, beta

    lagrange_multiplier = 0.0
    for i in range(maxiter):
        f = alpha - lagrange_multiplier/2 + 4*beta**2 / (2+lagrange_multiplier)**2 # Lagrangian
        df = -0.5 - 2*beta**2 / (2+lagrange_multiplier)**3 #Ableitung
        newton_step = f/df 
        lagrange_multiplier -= newton_step
        if abs(newton_step) < tol:
            break

    a = alpha - lagrange_multiplier*0.5
    b = 2*beta/(2+lagrange_multiplier)
    return a,b

def update_q(mesh, fes, phi, mu, r=0.1):
    # TODO das zerschießt mir die Dimensionen, spätestens in der zweiten Iteration -- Output Debuggen
    q_new = mu.CreateVector() # das nur bei GridFunction
    gphi = grad(phi)

    # Nach Definition von p_n in BB
    z = gphi + (1/r)*mu

    # Punktweise Minimierung
    for i in range(len(q_new)//2):

        alpha = z[0][i]
        beta  = z[1][i]

        a,b = newton(alpha,beta)

        q_new[2*i]   = a
        q_new[2*i+1] = b

    return q_new
    
```

## ALG Update in mu

```python
def update_mu(phi, q, mu, r): #Trivial?
    return (mu + r*(nabla_tx(phi) - q))
```

## ALG2 komplett

```python
def solve_ALG(r=0.1, maxiter=50):
    geo = SplineGeometry()

    # Eckpunkte
    p1 = geo.AppendPoint(0,0)
    p2 = geo.AppendPoint(1,0)
    p3 = geo.AppendPoint(1,1)
    p4 = geo.AppendPoint(0,1)

    ### Ränder definieren -- hier noch die Argumente nachvollziehen
    """
    Syntax: geo.Append([line segment from px to py] left/right domain, bc-name)
    """
    # Unterer Rand
    bottom = geo.Append(
        ["line", p1, p2],
        leftdomain=1,
        rightdomain=0,
        bc="bottom"
    )
    
    # Rechter Rand
    right = geo.Append(
        ["line", p2, p3],
        leftdomain=1,
        rightdomain=0,
        bc="right"
    )
    
    # Oberer Rand
    top = geo.Append(
        ["line", p3, p4],
        leftdomain=1,
        rightdomain=0,
        bc="top"
    )
    
    # Linker Rand = periodische Kopie von right
    left = geo.Append(
        ["line", p4, p1],
        leftdomain=1,
        rightdomain=0,
        copy=right,
        bc="left"
    )
    
    # Mesh aus dem Zeug da oben definieren
    mesh = Mesh(geo.GenerateMesh(maxh=0.2))


    # Periodische Randbedingungen in Raum; Neumann kommt scheinbar von selbst dazu
    #https://docu.ngsolve.org/latest/how_to/periodic.html#ngsolve.Periodic
    # H1 ist wohl Sobolev-Raum? order of finite element space; phase = None -> Periodic
    fes = Periodic(H1(mesh, order=2), phase = None) # phi(0,t) = phi(1,t)
    phi = GridFunction(fes)
    q = CoefficientFunction((x,y)) # Das hier doch als GridFunction definieren?
    mu = CoefficientFunction((x,y))
    for k in range(maxiter):
        phi = update_phi(mesh, fes, q, mu, r)
        #q = update_q(mesh, fes, phi, mu, r)# Das ändern, zeschießt die Dimensionen
        mu = update_mu(phi, q, mu, r)
    return phi, q, mu, mesh
    
*res,_ = solve_ALG()
#print(res)
for r in res:
    # Nette Bilder, aber machen die überhaupt Sinn?
    Draw(r, _, vectors={"grid_size":30})
```
